```python
from sklearn import linear_model
regr = linear_model.LinearRegression()
X = [[168, 0], [166, 0], [173, 0], [165, 0], [177, 0], [163, 0],
     [178, 0], [172, 0], [163, 1], [162, 1], [171, 1], [162, 1],
     [164, 1], [162, 1], [158, 1], [173, 1]]  # input data
y = [65, 61, 68, 63, 68, 61, 76, 67, 55, 51, 59, 53, 61, 56, 44, 57]

regr.fit(X, y)

print('coef :', regr.coef_)
print('intercept :', regr.intercept_)
print('regr_score :', regr.score(X, y))
print('Estimated weight :', regr.predict([[167, 0], [167, 1]]))
```

```python
#26_multi_Re_car.ipynb
import pandas as pd
from sklearn.linear_model import LinearRegression

#df = pd.read_csv("D:/01_data/03_data_car_1.csv")
df = pd.read_csv("https://raw.githubusercontent.com/victor4friend-eng/data_pub/refs/heads/main/03_data_car_1.csv")

X = df[['Volume','Weight']]
y = df['CO2']

model = LinearRegression()
model.fit(X,y)

print(model.coef_)
print(model.intercept_)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Volume, Weight 범위 생성
volume_range = np.linspace(df['Volume'].min(), df['Volume'].max(), 10)
weight_range = np.linspace(df['Weight'].min(), df['Weight'].max(), 10)

# 3D grid 생성
volume_grid, weight_grid = np.meshgrid(volume_range, weight_range)

# 회귀식으로 CO2 예측
co2_pred = (
    model.intercept_
    + model.coef_[0] * volume_grid
    + model.coef_[1] * weight_grid
)

# 3D 그래프
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# 실제 데이터
ax.scatter(
    df['Volume'],
    df['Weight'],
    df['CO2'],
    color='blue',
    label='Actual data'
)

# 회귀 평면
ax.plot_surface(
    volume_grid,
    weight_grid,
    co2_pred,
    color='orange',
    alpha=0.5
)

# 축 이름
ax.set_xlabel('Volume - Abduqodir')
ax.set_ylabel('Weight')
ax.set_zlabel('CO2')

ax.set_title('Multiple Linear Regression: Volume & Weight vs CO2')

plt.show()
```
